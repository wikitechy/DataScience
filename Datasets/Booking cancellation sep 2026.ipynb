```python
from sklearn.model_selection import GridSearchCV,StratifiedKFold
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score,f1_score
```

```python
def tuning(cls,params):
    cv = StratifiedKFold(n_splits=5,shuffle=True)
    class_grid = GridSearchCV(cls,params,cv=cv,scoring='f1',n_jobs=-1)
    class_grid.fit(X_train,y_train)
    best_hp = class_grid.best_params_
    return class_grid.best_estimator_,best_hp
```

```python
dt = DecisionTreeClassifier()
```

```python
dt_parameter_grid = {
    'criterion':['gini','entropy'],
    'max_depth':[12,15,20]
}
```

```python
tuning(dt,dt_parameter_grid)
```

```python
dt = DecisionTreeClassifier(criterion= 'gini', max_depth= 20)
```

```python
dt.fit(X_train,y_train)
```

```python
y_pred = dt.predict(X_test)
```

```python
accuracy_score(y_test,y_pred)
```
