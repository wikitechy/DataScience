```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"<.*?>"," ",text)
    text = re.sub(r"[^a-zA-Z\s]"," ",text)
    text = re.sub(r"\s+"," ",text)
    x = []
    a = stopwords.words('english')
    for i in text.split():
        if i not in a:
            x.append(i)
    for i in range(len(x)):
        x[i] = (ps.stem(x[i]))
    
    text = " ".join(x)
    return text
```

# Predict next word - RNN

```python
import os, re, json, numpy as np, tensorflow as tf 
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras import layers, Sequential
from pathlib import Path
```

```python
DATA_PATH = Path('shakespeare.txt')

text = DATA_PATH.read_text(encoding='utf-8-sig')[:1000000]
```

```python
text = re.sub(r"(?m)^\s*[A-Z][A-Z ]{1,25}\.\s*",' ',text)
```

```python
tokennizer = Tokenizer(oov_token='<OOV>')
```

```python
tokennizer.fit_on_texts([' '.join(words)])
```

```python
vocab_size = len(tokennizer.word_index)+1
```

```python
vocab_size
```

```python
ids = tokennizer.texts_to_sequences([' '.join(words)])[0]
```

```python
ids
```

```python
window,sequences = 6, []
```

```python
for i in range(window,len(ids)):
    sequences.append(ids[i-window:i+1])
```

```python
sequences = np.array(sequences,dtype=np.int32)
```

```python
model = Sequential([
    layers.Embedding(vocab_size,32),
    layers.RNN(48),
    layers.Dense(vocab_size,activation='softmax')
])
```

```python
index_to_word = {i:w for w,i in tokennizer.word_index.items()}
```

```python
index_to_word
```

```python
def predict_next_word(prompt,top_k=1):
    ids = tokennizer.texts_to_sequences([prompt.lower()])[0][-window:]
    ids = pad_sequences([ids],maxlen=window,padding='pre')
    probabilities = model.predict(ids,verbose=0)[0]
    best = np.argsort(probabilities)[-top_k:][::-1]
    return [(index_to_word.get(int(i),'<unknown>'),round(float(probabilities[i]),4)) for i in best]
```

```python
prompt = 'the king is the lord'
print('Prompt:',prompt)
for word,probability in predict_next_word(prompt):
    print(f"{word:15} {probability:.2%}")
```

```python
import os 
import random
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import torch

from datasets import load_dataset,Dataset
from sklearn.model_selection import train_test_split
from sklearn.metrics import (accuracy_score,
precision_recall_fscore_support,
classification_report,confusion_matrix
)
from transformers import (AutoTokenizer, 
AutoModelForSequenceClassification, TrainingArguments, Trainer
)
```

```python
SEED = 42
```

```python
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
```

```python
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print('Device:',DEVICE)
if torch.cuda.is_available():
    print('GPU:',torch.cuda.get_device_name(0))
```

```python
TRAIN_SAMPLES = 2000
VALIDATION_SAMPLES = 500
TEST_SAMPLES = 1000
MAX_LENGTH = 120
BATCH_SIZE = 8
EPOCHS = 2
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
```

```python
dataset = load_dataset("stanfordnlp/imdb")
```

```python
dataset
```

```python
dataset['train'][0]
```

```python
train_pool = pd.DataFrame(dataset['train']).sample(
    n=min(TRAIN_SAMPLES+VALIDATION_SAMPLES,len(dataset['train'])),random_state=SEED)
```

```python
train_pool = train_pool.reset_index(drop=True)
```

```python
train_pool
```

```python
test_df = pd.DataFrame(dataset['test']).sample(
    n=min(TEST_SAMPLES,len(dataset['test'])),
    random_state=SEED
    ).reset_index(drop=True)
```

```python
train_df,val_df = train_test_split(train_pool,
                                  test_size=VALIDATION_SAMPLES/len(train_pool),
                                  random_state=SEED,
                                  stratify=train_pool['label'])

```

```python
train_df.shape
```

```python
val_df.shape
```

```python
test_df.shape
```

```python
MODEL_NAME = 'distilbert/distilbert-base-uncased'
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
sample_token = tokenizer('the movie was fantastic')
sample_token
```

```python
def tokenize_function(batch):
    return tokenizer(batch['text'],truncation=True,max_length=MAX_LENGTH,padding='max_length')
```

```python
train_hf = Dataset.from_pandas(train_df[['text','label']],preserve_index=False)
```

```python
train_hf
```

```python
test_hf = Dataset.from_pandas(test_df[['text','label']],preserve_index=False)
val_hf = Dataset.from_pandas(val_df[['text','label']],preserve_index=False)
```

```python
train_tokenized = train_hf.map(tokenize_function,batched=True)
val_tokenized = val_hf.map(tokenize_function,batched=True)
test_tokenized = test_hf.map(tokenize_function,batched=True)
```

```python
train_tokenized[0]
```

```python
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME,
                                                          num_labels = 2,
                                                          id2label = {0:'Negative',
                                                                      1:'Positive'},
                                                          label2id = {'Negative':0,
                                                                      'Positive':1})
```

```python
def compute_metrics(eval_pred):
    logits,labels = eval_pred
    prediction = np.argmax(logits,axis=-1)
    precision,recall,f1,_ = precision_recall_fscore_support(labels,prediction,
                                                            average='binary',
                                                           zero_division=0)

    return {
        'accuracy':accuracy_score(labels,prediction),
        'precision': precision,
        'recall':recall,
        'f1':f1
    }
```

```python
training_args = TrainingArguments(
    output_dir='./sentiment_results',
    eval_strategy='epoch',
    save_strategy='epoch',
    learning_rate=LEARNING_RATE,
    per_device_eval_batch_size=BATCH_SIZE,
    per_device_train_batch_size=BATCH_SIZE,
    num_train_epochs=EPOCHS,
    weight_decay=WEIGHT_DECAY,
    load_best_model_at_end=True,
    metric_for_best_model='f1',
    greater_is_better=True,
    logging_steps=25,
    report_to='none',
    seed = SEED
)
```

```python
trainer = Trainer(
    model = model,
    args= training_args,
    train_dataset=train_tokenized,
    eval_dataset=val_tokenized,
    compute_metrics=compute_metrics
)
```

```python
train_output = trainer.train()
```

```python
test_result = trainer.evaluate(test_tokenized)
```

```python
pred_output = trainer.predict(test_tokenized)
test_logits = pred_output.predictions
test_labels = pred_output.label_ids

test_prediction = np.argmax(test_logits,axis=-1)

print(classification_report(test_labels,test_prediction,target_names=['Negative','Positive'],
                           zero_division=0))

cm = confusion_matrix(test_labels,test_prediction,labels=[0,1])

cm
```

```python
def predict_new_sentiment():
    review = input('Review:')
    trainer.model.eval()
    inputs = tokenizer(
        review,return_tensors = 'pt',
        truncation=True,max_length=MAX_LENGTH,padding=True
    )

    inputs = {k:v.to(trainer.model.device) for k,v in inputs.items()}

    with torch.no_grad():
        outputs = trainer.model(**inputs)

    proba = torch.softmax(outputs.logits,dim=-1)[0]

    predicted_class = int(torch.argmax(proba).item())
    score = float(proba[predicted_class].item())

    return {
        'review':review,
        'sentiment':'Positive' if predicted_class ==1 else 'Negative',
        'score': round(score,4)
    }
    
```
